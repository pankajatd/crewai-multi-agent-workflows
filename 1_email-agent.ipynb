```python
import os
from dotenv import load_dotenv
from crewai import LLM

# 1. Load your NEW API key
load_dotenv()

# 2. Use the 2026 Stable Model String
# Note the "gemini/" prefix and the "2.5" version
llm = LLM(
    model="gemini/gemini-2.5-flash",
    temperature=0.1
)

print("Attempting to connect to Gemini 2.5 Flash...")

try:
    # 3. Test the call
    response = llm.call("Who invented transcendental meditation?")
    print("-" * 30)
    print(f"✅ Success! Answer: {response}")
    print("-" * 30)
except Exception as e:
    # If 2.5-flash is busy, we try the newest 3.0-flash
    print(f"Direct call failed, trying fallback to Gemini 3...")
    try:
        llm_fallback = LLM(model="gemini/gemini-3-flash", temperature=0.1)
        response = llm_fallback.call("Who invented transcendental meditation?")
        print(f"✅ Success with Gemini 3! Answer: {response}")
    except Exception as e2:
        print(f"❌ Critical Error: {e2}")
```

Creating an Email Assistant Agent to Write the Email in a Professional way.

Here we are providing the original Email to the Email Assistant Agent,Provide the BackStory and LLM object.
We specify the task as :
1.Take the following rough email and rewrite into  a professional polished
     version.
2.  Expand the Abbervations.
     and the expected_output from the Email_Assistant_Agent "A Professional written email with proper formatting and content."

```python
from crewai import Agent,Task,Crew
email_assistant=Agent(
    role="Email Assistant Agent",
    goal="Improve emails and make them sound professional and clear",
    backstory="A highly experienced communication expert skilled in professional email writing",
    verbose=True,
    llm=llm
)

original_email="""
            hey team,just wanted to tell you that the demo is kind of ready,but there is still stuff left.
            May we can show what we have and say reset is WIP
            let me know what you think.Thanks
           """
email_task=Task(

     description=f"""Take the following rough email and rewrite into  a professional polished
     version.
     Expand abbervations:
     '''{original_email}'''""",
    agent=email_assistant,
    expected_output="A Professional written email with proper formatting and content.",
)

crew=Crew(
     agents=[email_assistant],
     tasks=[email_task],
    verbose=True,

)

result = crew.kickoff()
print(result)
```
