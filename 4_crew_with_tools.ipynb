This Code is to create crew of Agents that interact with each other.**Reearch Agent, Writer Agent and Serper tool** to access Data from the Internet.

**Research Agent**-Does Reserach on the specified Topic using Serper tool to get the latest Data and provides the Result.
**Writer Agent**-This Agent Receives the Date from the Research Agent and writes a proper Blog.


**Creating a Research Agent and a Writer Agnet**
Research Agent=Will do the Research on the defined Topic and Provide the data to the Writer Agent.
In order to get the Latest Data for the specified Topic,Research Agent uses the SerPer Tool to connect to the Internet and find the Latest data
Writer Agent= Will take the Data provided by the Writer agent and Write a Blog.

```python
import os
from dotenv import load_dotenv
from crewai import LLM, Agent, Task, Crew
from crewai_tools import SerperDevTool

# 1. Load your API keys (.env should have GEMINI_API_KEY and SERPER_API_KEY)
load_dotenv()

# 2. Setup the LLM
llm = LLM(
    model="gemini/gemini-2.5-flash",
    temperature=0.1
)

# 3. Define Agents
research_agent = Agent(
       role="Research Specialist",
       goal="Research interesting facts about the topic: {topic}",
       backstory="You are an expert at finding relevant and factual data",
       tools=[SerperDevTool()], # FIXED: Added parentheses here
       verbose=True,
       llm=llm
)

writer_agent = Agent(
    role="Writer",
    goal="Write interesting facts about the topic: {topic}", # FIXED: Typo 'Writer' to 'Write'
    backstory="You are an expert at crafting engaging narratives from data.",
    llm=llm,
    verbose=True,
)
```

```python
task1=Task(
    description='Find 3-5 interesting and recent facts about the topic: {topic} as of year 2025',
    expected_output='A bullet list of 3-5 facts about the topic: {topic}',
    agent=research_agent,
)

task2=Task(
    description='Write a 100-word blog post summary about topic: {topic} using the facts from the research.',
    expected_output='A blog Post Summary about topic: {topic}',
    agent=writer_agent,
    context=[task1],
)
```

```python
crew=Crew(

    agents=[research_agent, writer_agent],
    tasks=[task1, task2],
    verbose=True,
)


crew=crew.kickoff({'topic':"The Future of Electrical Vehicles"})
```
